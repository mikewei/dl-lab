```python
import gymnasium as gym
from gymnasium.vector import VectorObservationWrapper
from gymnasium.wrappers.vector import RescaleAction
import numpy as np
import torch
from torch import nn, optim
from torch.nn import functional as F
import optuna
import multiprocessing
from dlbase.funcs.decay import ExpDecay
from dlbase.funcs.avg import EWMA
from dlbase.funcs.mapping import rescale_tensor
import dlbase.rl.gyms as gyms
import dlbase.rl.algo as algo
import importlib
importlib.reload(gyms)
importlib.reload(algo)

class CarRacingEnvs(VectorObservationWrapper):
    def __init__(self, num_envs: int, **kwargs):
        rescaled_envs = RescaleAction(gym.make_vec('CarRacing-v3', num_envs=num_envs, render_mode='rgb_array', **kwargs),
                                     min_action=-1.0, max_action=1.0)
        super().__init__(rescaled_envs)
    def observations(self, observations: np.ndarray):
        assert observations.shape == (self.num_envs, 96, 96, 3)
        v = torch.tensor([car.hull.linearVelocity for car in self.unwrapped.get_attr('car')])  # type: ignore
        v = rescale_tensor(v, (0, 255), (-60, 600)).to(torch.uint8)
        observations[:, 95, 95, :2] = v
        return observations

def vgg_block(num_convs, out_channels):
    layers = []
    for _ in range(num_convs):
        layers.append(nn.LazyConv2d(out_channels, 3, padding=1))
        layers.append(nn.ReLU())
    layers.append(nn.MaxPool2d(2, stride=2))
    return nn.Sequential(*layers)

def init_weights(module: nn.Module):
    for m in module.modules():
        if isinstance(m, nn.Conv2d):
            nn.init.kaiming_normal_(m.weight, mode='fan_out', nonlinearity='relu')
            if m.bias is not None:
                nn.init.constant_(m.bias, 0)
        elif isinstance(m, nn.Linear):
            nn.init.kaiming_normal_(m.weight, mode='fan_in', nonlinearity='relu')
            if m.bias is not None:
                nn.init.constant_(m.bias, 0)

vgg_arch = [(1,16),(1,32),(1,64),(1,128)]

class CarRacingCNN(nn.Module):
    def __init__(self, arch=vgg_arch, hidden_dim=128):
        super().__init__()
        conv_blks = []
        for (num_convs, out_channels) in arch:
            conv_blks.append(vgg_block(num_convs, out_channels))
        self.net = nn.Sequential(*conv_blks, nn.Flatten(),
                                 nn.LazyLinear(hidden_dim * 2), nn.ReLU(),
                                 nn.LazyLinear(hidden_dim))
        with torch.no_grad():
            self(torch.randn(1, 96, 96, 3))
            init_weights(self)
    def forward(self, s: torch.Tensor):
        assert s.shape[1:] == (96, 96, 3)
        s = s.transpose(1, 3) / 255.0
        return self.net(s)

class CarRacingPPOActor(algo.PActor):
    cnn: nn.Module
    def __init__(self, cnn: nn.Module, hidden_dim=256):
        super().__init__()
        object.__setattr__(self, 'cnn', cnn)
        self.fc1 = nn.Sequential(nn.LazyLinear(hidden_dim), nn.ReLU())
        self.output = nn.Sequential(nn.LazyLinear(3), nn.Tanh())
        self.std = nn.Parameter(torch.ones(3) * 0.5)
        with torch.no_grad():
            self(torch.randn(1, 96, 96, 3))
            init_weights(self)
        # pretrain
        o = optim.Adam(self.parameters(), lr=1e-3)
        for _ in range(500):
            s = torch.randn(128, 96, 96, 3)
            loss = F.mse_loss(self(s)[0], torch.tensor([[0.0, 0.6, -1.0]]))
            o.zero_grad()
            loss.backward()
            o.step()

    def forward(self, s: torch.Tensor):
        with torch.no_grad():
            v = s[:, 95, 95, :2]
            s[:, 95, 95, :2] = 0
            sf = self.cnn(s)
        a = self.output(self.fc1(torch.cat([sf, v], dim=1)))
        return a, self.std.expand_as(a)

class CarRacingPPOCritic(algo.VCritic):
    def __init__(self, hidden_dim=1024):
        super().__init__()
        self.cnn = CarRacingCNN()
        self.fc1 = nn.Sequential(nn.LazyLinear(hidden_dim), nn.ReLU())
        # self.fc2 = nn.Sequential(nn.LazyLinear(hidden_dim))
        self.output = nn.LazyLinear(1)
        with torch.no_grad():
            self(torch.randn(1, 96, 96, 3))
            init_weights(self)
    def forward(self, s: torch.Tensor):
        with torch.no_grad():
            v = s[:, 95, 95, :2]
            s[:, 95, 95, :2] = 0
        q = self.fc1(torch.cat([self.cnn(s), v], dim=1))
        # q = F.relu((self.fc2(q) + q))
        return self.output(q).squeeze(-1)

def model_factory():
    critic = CarRacingPPOCritic()
    actor = CarRacingPPOActor(critic.cnn)
    return actor, critic




# 定义目标函数
def objective(trial):
    lr=trial.suggest_float("lr", 1e-5, 1e-2, log=True)
    gamma=trial.suggest_float("gamma", 0.9, 0.999, log=True)
    c2=trial.suggest_float("c2", 1e-5, 1e-2, log=True)
    regu_p = trial.suggest_float("regu_p", 1e-5, 1e-2, log=True)
    env = CarRacingEnvs(num_envs=4)
    actor, critic = model_factory()
    ppo = algo.PPO(actor, critic, True, lr=lr, gamma=gamma, lambd=0.95, eps_clip=0.3, train_per_steps=1000, c2=c2,
                action_regu_fn=lambda x: (0.01 * x[..., 0]**2 + regu_p * (x[..., 1]-1)**2 + regu_p * (x[..., 2]+1)**2).mean())
    ewma = EWMA(0.99)
    runner = gyms.GymRunner(env, ppo, show_env_freq=0, max_episodes=400, memory_size=200000, 
                            stop_on_rewards=lambda r: ewma(r) > 850, with_old_memory=None)
    runner.run()
    return ewma.value

# 定义优化函数
def optimize_with_map(storage_url_study_name):
    storage_url, study_name = storage_url_study_name
    study = optuna.load_study(study_name=study_name, storage=storage_url)
    study.optimize(objective, n_trials=10)  # 每个进程优化10次

if __name__ == '__main__':
    # 配置数据库存储
    storage_url = "sqlite:///car_racing_ppo_hpo.db"
    study_name = "car_racing_ppo_hpo_5"

    # 创建 Study
    study = optuna.create_study(study_name=study_name, storage=storage_url, direction="maximize")
    # study.optimize(objective, n_trials=10)

    # 使用 Pool.map
    num_processes = 2  # 定义进程数量
    with multiprocessing.Pool(processes=num_processes) as pool:
        # 将进程任务分配到不同进程，每个任务使用相同的存储和 Study 名称
        pool.map(optimize_with_map, [(storage_url, study_name)] * num_processes)

    # 查看最佳结果
    print("Best parameters:", study.best_params)
    print("Best value:", study.best_value)
```
